```python
from scipy import stats

def prepare_data(df):
    # Удаление колонок с большим количеством пропусков
    df = df.drop(['osm_accommodation_closest', 'osm_accommodation_closest_5_mean'], axis=1)
    
    for cl in df.columns:
        # Замена бесконечных значений
        df[cl].replace([np.inf, -np.inf], np.nan)
        
        if df[cl].dtype in ['object', 'category']:
            # Для категориальных - мода
            df[cl] = df[cl].fillna(df[cl].mode()[0])
        elif df[cl].dtype in ['int64', 'float64']:
            # Для числовых - медиана
            df[cl] = df[cl].fillna(df[cl].median())
            
            # Удаление выбросов методом z-score
            z = np.abs(stats.zscore(df[cl]))
            median = df[cl].median()
            df.loc[z > 3, cl] = median
    
    return df

# Применение ко всем данным
train_merged = prepare_data(train_merged)
test_merged = prepare_data(test_merged)
```

```python
def transaction(data, n_neighbours, is_train=True):
    df = data.copy()
    df['transaction_sum_knn'] = np.nan
    
    # Разные KNN модели для каждого типа бизнеса
    groups = df.groupby("industry_type")
    base_df = train_merged if not is_train else df
    
    for industry_type, group in groups:
        knn = NearestNeighbors(n_neighbors=n_neighbours, algorithm='ball_tree', metric='haversine')
        knn.fit(np.radians(base_df.loc[base_df['industry_type'] == industry_type, ['lat', 'lon']].to_numpy()))
        
        distances, indices = knn.kneighbors(np.radians(group[['lat', 'lon']]).to_numpy())
        
        # Для обучающих данных исключаем саму точку, для тестовых - нет
        first_idx = 1 if is_train else 0       
        transaction_sum_pred = [
            base_df.iloc[idx[first_idx:]]['customer_count'].mean()
            for idx in indices
        ]
        df.loc[group.index, 'transaction_sum_knn'] = transaction_sum_pred
    
    return df

# Создание признаков
train_merged = transaction(train_merged, n_neighbours=5, is_train=True)
test_merged = transaction(test_merged, n_neighbours=5, is_train=False)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from math import sqrt


X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=train_merged['industry_type']
)

X_train, X_test, y_train, y_test = train_test_split(
    X_train, y_train, test_size=0.4, random_state=42, stratify=X_train['industry_type']
)
```

```python
import optuna
import lightgbm as lgb
import numpy as np
import pandas as pd
def objective(trial):
    params = {
        'objective': 'regression',
        'metric': 'mae',
        'boosting_type': 'gbdt',
        'verbosity': -1,
        'seed': 42,
        'num_leaves': trial.suggest_int('num_leaves', 20, 128),
        'max_depth': trial.suggest_int('max_depth', 3, 12),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'min_data_in_leaf': trial.suggest_int('min_data_in_leaf', 20, 100),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.6, 1.0),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.6, 1.0),
        'bagging_freq': trial.suggest_int('bagging_freq', 1, 5)
    } 
    model = LGBMRegressor(**params)
    model.fit(
        X_train, y_train,
        eval_set = [(X_valid, y_valid)]
    )
    preds = model.predict(X_valid)
    MAE = mean_absolute_error(y_valid, preds)
    return MAE

study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=100)
#Выполняем подбор параметров при помощи optuna к LGBMRegressor.
```

```python

final_model = LGBMRegressor(
    **study.best_params, 
    verbose = 1
)
# Обучаем нашу модель на подобранных парамтов
final_model.fit(
    pd.concat([X_train, X_valid], ignore_index = True),
    pd.concat([y_train, y_valid], ignore_index = True)
)

y_pred = final_model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_pred)
print(f"MAE: {mae:.4f}  Score: {1 / (1 + mae):.4f}") # Проверяем, какой бы примерный балл нам бы вернула система.
```
