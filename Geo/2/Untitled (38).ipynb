```python
train_data['price_target'] = np.log10(train_data['price_target'])
print(train_data['price_target'][:10])

# Комментарий: ВАЖНОЕ ПРЕОБРАЗОВАНИЕ! Логарифмирование целевой переменной по основанию 10. 
# Это стандартный прием для задач регрессии с большим разбросом данных, помогает модели лучше обучаться
```

```python
price_mean_hc = train_data.groupby("hc_name_cat")['price_target'].mean().reset_index()

print("1")

train_data = train_data.merge(
    price_mean_hc,
    on = "hc_name_cat",
    how="left",
    suffixes=('_left', '_right')
)
#train_data.head(30)
train_data.rename(columns={"price_target_left": 'price_target', 'price_target_right': 'hc_cat_mean_price'}, inplace=True)

# Комментарий: ФИЧИНЖИНИРИНГ! Создание новой фичи - средняя цена по категории недвижимости.
# Это помогает модели учитывать "престижность" или типичную стоимость объектов определенной категории
```

```python
train_data
```

```python

test_data = test_data.merge(
    price_mean_hc,
    on = "hc_name_cat",
    how="left",
    suffixes=('_left', '_right')
)

test_data.rename(columns={'price_target': 'hc_cat_mean_price'}, inplace=True)

# Комментарий: Применяем то же преобразование к тестовым данным - добавляем среднюю цену по категориям
```

```python
from sklearn.preprocessing import OneHotEncoder

def prepare_data(df, name ):
    print(f"{50 * '-'}\nStart preprocessing {name}\n{50 * '-'}")
    # for cl in df.columns:
    #     if df[cl].isna().sum() > 0:
    #         print(f"Nan's in {cl} : {df[cl].isna().sum()}")
    for cl in df.columns:
        if (df[cl].isna().sum() > 0):
            if (df[cl].dtype in ['object', 'category']):
                df[cl] = df[cl].fillna(df[cl].mode[0])
            else :
                df[cl] = df[cl].fillna(df[cl].median())
    for cl in df.columns:
        if list(df[cl]).count(-999.000000)  > 0:
            df[cl] = df[cl].replace(-999.000000, df[cl].median())
            print(f"count -999's in {cl} : {list(df[cl]).count(-999.000000)}")
    
    for cl in df.columns:
        if df[cl].isna().sum() > 0:
            print(f"Nan's in {cl} : {df[cl].isna().sum()}")
    print(f"End of preprocessing {name}")
    return df
train_data = prepare_data(train_data, "train_data")
test_data = prepare_data(test_data, "test_data")

ohe_cols = ['rooms_4', 'region_name_cat']
# for cl in set(train_data.columns) & set(test_data.columns):
#     if train_data[cl].dtype == "category":
#     #if len(train_data[cl].unique()) < 11 or len(test_data[cl].unique()) < 11:
#         print(f"OneHot in {cl} : {train_data[cl].unique()}")
#         ohe_cols.append(cl)

ohe = OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore')

train_ohe = ohe.fit_transform(train_data[ohe_cols])
test_ohe = ohe.transform(test_data[ohe_cols])

train_ohe_df = pd.DataFrame(train_ohe, columns=ohe.get_feature_names_out(ohe_cols), index=train_data.index)
test_ohe_df = pd.DataFrame(test_ohe, columns=ohe.get_feature_names_out(ohe_cols), index=test_data.index)


train_data = pd.concat([train_data.drop(columns=ohe_cols), train_ohe_df], axis=1)
test_data = pd.concat([test_data.drop(columns=ohe_cols), test_ohe_df], axis=1)

# print(train_data.head())
# print(test_data.head())

# Комментарий: ПРЕОБРАЗОВАНИЕ ДАННЫХ! 
# 1. Обработка пропусков: для категориальных - мода, для числовых - медиана
# 2. Замена специального значения -999 на медиану (это явно пропуски в данных)
# 3. One-Hot Encoding для категориальных признаков 'rooms_4' и 'region_name_cat'
#    Преобразует категории в бинарные столбцы для работы алгоритмов машинного обучения
```

```python
train_data['agreement_date'] = pd.to_datetime(train_data['agreement_date'])

train_data['year'] = train_data['agreement_date'].dt.year
train_data['month'] = train_data['agreement_date'].dt.month
train_data['day'] = train_data['agreement_date'].dt.day
train_data = train_data.drop('agreement_date', axis = 1)
#print(train_data.head())

test_data['agreement_date'] = pd.to_datetime(test_data['agreement_date'])

test_data['year'] = test_data['agreement_date'].dt.year
test_data['month'] = test_data['agreement_date'].dt.month
test_data['day'] = test_data['agreement_date'].dt.day
test_data = test_data.drop('agreement_date', axis = 1)
#print(train_data.head())

# Комментарий: РАБОТА С ДАТАМИ! Разбиваем дату на год, месяц и день как отдельные признаки.
# Это помогает модели учитывать сезонность и временные тренды в ценах на недвижимость
```

```python
from catboost import CatBoostRegressor, Pool
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

X = train_data.drop(columns=['price_target'])
y = train_data['price_target']


X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = CatBoostRegressor(
    iterations=7000,
    learning_rate=0.025,
    depth=9,  
    eval_metric='MAE',
    random_seed=42,
    verbose=400
)
model.fit(
    X_train, y_train,
    eval_set=(X_val, y_val),
    #cat_features=cat_features,
    early_stopping_rounds=50
)
y_pred = model.predict(X_val)
acc = mean_absolute_error(y_val, y_pred)
print(f"MAE: {acc:.4f}")

# Комментарий: ОСНОВНОЕ МОДЕЛИРОВАНИЕ!
# Используем CatBoost - мощный алгоритм для табличных данных
# - 7000 итераций: много, но есть early stopping
# - learning_rate=0.025: умеренная скорость обучения
# - depth=9: достаточно глубокая модель для сложных зависимостей
# - MAE метрика: соответствует метрике соревнования (MAPE)
# - 30% данных на валидацию для контроля переобучения
```
